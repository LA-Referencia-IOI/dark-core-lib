# dARK Core Lib - Read-Only Demo

This notebook demonstrates observer-style usage with `dark-core-lib` in read-only mode.


## 1. Setup

Expected env file at project root (`../.env.integration` preferred).


```python
import os
from dotenv import load_dotenv

if os.path.exists('../.env.integration'):
    load_dotenv('../.env.integration')
    print('Loaded ../.env.integration')
elif os.path.exists('../.env'):
    load_dotenv('../.env')
    print('Loaded ../.env')
else:
    print('No env file found. Create ../.env.integration or ../.env')
```

## 2. Initialize Read-Only Client


```python
from dark_core_lib import DARKCoreClient

client = DARKCoreClient.from_env(read_only=True)
print('Connected:', client.is_connected())
print('Block:', client.get_block_number())
```

## 3. Resolve / Fetch ARKs

Set a known ARK from your local network before running these cells.


```python
NAAN = '12345'
NAME = 'test-document'  # change if needed

exists = client.ark_exists(NAAN, NAME)
print(f'Exists ark:/{NAAN}/{NAME}:', exists)
```

```python
if client.ark_exists(NAAN, NAME):
    url = client.resolve_ark(NAAN, NAME)
    info = client.get_ark(NAAN, NAME)
    print('Resolved URL:', url)
    print('ARK ID:', info.ark_id)
    print('Owner:', info.owner)
    print('CID:', info.cid)
else:
    print('ARK not found. Create one first with the E2E notebook.')
```
